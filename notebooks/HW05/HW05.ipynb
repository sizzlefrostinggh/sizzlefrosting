# 7 Dropout

What is dropout in the context of a neural network and how is it beneficial in training a neural network?

During training, a certain percentage of neurons in a hidden layer are randomly dropped/deactivated.  Those dropped neurons do not participate in forward propogation, contribute to the loss calculation, or receive gradient updates during backpropogation for that iteration.

This is beneficial in multiple ways.  It helps prevent overfitting by making the network not rely too much on particular neurons.  It helps reduce neuron co-adaptation making them learn features more independently.  It helps improve generalization so it can perform better on unseen data.  Finally, it basically trains multiplpe subnetworks that share weights and their combined behavior can improve predictions.

# 8 Principal Component Analysis on Diabetes Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Load the Diabetes dataset
url = "https://raw.githubusercontent.com/rahulbhadani/Datasets/master/Diabetes/diabetes.csv"
df = pd.read_csv(url)
print(df.head())

X = df.drop(columns=["Outcome"]).values
y = df["Outcome"].values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 1.  Principal Component Analysis (PCA) with Two Components
pca2 = PCA(n_components=2)
X_pca2 = pca2.fit_transform(X_scaled)

print("\n2-Component PCA Explained Variance")
print(f"PC1 Explained Variance : {pca2.explained_variance_ratio_[0]:.4f} ({pca2.explained_variance_ratio_[0]*100:.2f}%)")
print(f"PC2 Explained Variance : {pca2.explained_variance_ratio_[1]:.4f} ({pca2.explained_variance_ratio_[1]*100:.2f}%)")
print(f"Total Variance (2D)    : {np.sum(pca2.explained_variance_ratio_):.4f} ({np.sum(pca2.explained_variance_ratio_)*100:.2f}%)")

# 2.  2D Plot Analysis
plt.figure(figsize=(8, 6))
scatter = plt.scatter(X_pca2[:, 0], X_pca2[:, 1], c=y, cmap="coolwarm", alpha=0.6, edgecolors="k", s=30)
plt.xlabel(f"PC1 ({pca2.explained_variance_ratio_[0]*100:.2f}%)")
plt.ylabel(f"PC2 ({pca2.explained_variance_ratio_[1]*100:.2f}%)")
plt.title("2D PCA of Diabetes Dataset")
plt.legend(handles=scatter.legend_elements()[0], labels=["No Diabetes (0)", "Diabetes (1)"])
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()

print("\nLooking at the 2D plot the datasetsare not distinguishable.  There is too much overlap between the to be able to distinguish them from each other.\n")

# 3.  PCA with Three Components
pca3 = PCA(n_components=3)
X_pca3 = pca3.fit_transform(X_scaled)

print("3-Component PCA Explained Variance")
for i, var in enumerate(pca3.explained_variance_ratio_, 1):
    print(f"PC{i} Explained Variance : {var:.4f} ({var*100:.2f}%)")
print(f"Total Variance (3D)    : {np.sum(pca3.explained_variance_ratio_):.4f} ({np.sum(pca3.explained_variance_ratio_)*100:.2f}%)")

print("\nAdding the third component adds an additional 12.87% of variance, rasing the total variance from 47.82% to 60.69%.\n")

# 4.  3D Plot Analysis
fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection="3d")
sc3 = ax.scatter(X_pca3[:, 0], X_pca3[:, 1], X_pca3[:, 2], c=y, cmap="coolwarm", alpha=0.6, edgecolors="k", s=25)
ax.set_xlabel(f"PC1 ({pca3.explained_variance_ratio_[0]*100:.2f}%)")
ax.set_ylabel(f"PC2 ({pca3.explained_variance_ratio_[1]*100:.2f}%)")
ax.set_zlabel(f"PC3 ({pca3.explained_variance_ratio_[2]*100:.2f}%)")
ax.set_title("3D PCA of Diabetes Dataset")
ax.legend(handles=sc3.legend_elements()[0], labels=["No Diabetes (0)", "Diabetes (1)"])
plt.show()

print("\nAdding the third component spreads out the points more but there is still so much overlap that it still just as distinguishable as it was in the 2D plot without PC3.\n")
```

# 9 Transforming Data with Kernels and SVM

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# 1.  Generate synthetic circles dataset
X, y = make_circles(n_samples=400, noise=0.1, factor=0.3, random_state=1)

plt.figure(figsize=(6, 5))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", edgecolors="k", s=30)
plt.title("Synthetic Circles Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()

print("\nA linear SVM will fail here because the two classes form rings and a straight line wouldn't be able to separate the inner circle from the ring that is surrounding it.\n")

def plot_decision_boundary(model, X, y, title):
    x_min, x_max = X[:, 0].min() - 0.3, X[:, 0].max() + 0.3
    y_min, y_max = X[:, 1].min() - 0.3, X[:, 1].max() + 0.3
    xx, yy = np.meshgrid(np.linspace(x_min, x_max, 300), np.linspace(y_min, y_max, 300))
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

    plt.figure(figsize=(6, 5))
    plt.contourf(xx, yy, Z, cmap="coolwarm", alpha=0.3)
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", edgecolors="k", s=30)
    plt.title(title)
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")
    plt.grid(True, linestyle="--", alpha=0.5)
    plt.show()

# 2.  Implement a Linear SVM classifier
svm_linear = SVC(kernel="linear", random_state=1)
svm_linear.fit(X, y)
acc_linear = accuracy_score(y, svm_linear.predict(X))
print(f"Linear SVM Accuracy: {acc_linear * 100:.2f}%")
plot_decision_boundary(svm_linear, X, y, f"Linear SVM (Accuracy: {acc_linear*100:.2f}%)")

print(f"\nThe accuracy being {acc_linear*100:.2f}% shows us that a straight line isn't good splitting the ring and circle and that it is almost the same as just flipping a coin and guessing.\n")

# 3.  Implement non-linear SVM using the RBF kernel with gamma = 1.0
svm_rbf = SVC(kernel="rbf", gamma=1.0, random_state=1)
svm_rbf.fit(X, y)
acc_rbf = accuracy_score(y, svm_rbf.predict(X))
print(f"RBF SVM (gamma=1.0) Accuracy: {acc_rbf * 100:.2f}%")
plot_decision_boundary(svm_rbf, X, y, f"RBF SVM, gamma=1.0 (Accuracy: {acc_rbf*100:.2f}%)")

print(f"\nThe RBF kernel maps the 2D points correctly, producing a circular decision boundary in 2D with {acc_rbf*100:.2f}% accuracy.\n")

# 4.  RBF gamma paramenter
for g in [0.1, 10.0]:
    svm_g = SVC(kernel="rbf", gamma=g, random_state=1)
    svm_g.fit(X, y)
    acc_g = accuracy_score(y, svm_g.predict(X))
    print(f"RBF SVM (gamma={g}) Accuracy: {acc_g * 100:.2f}%")
    plot_decision_boundary(svm_g, X, y, f"RBF SVM, gamma={g} (Accuracy: {acc_g*100:.2f}%)")

print("\nA small gamma (0.1) makes a wider radius which makes a larger smoother boundry that could have a high bias and underfit data.")
print("A large gamma (10.0) makes a smaller raidus which makes a smaller tight boundary that could have high variance and overfit unseen data.\n")
```

# 10 K-means

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.cluster import KMeans
from sklearn.metrics import (
    adjusted_rand_score,
    v_measure_score,
    silhouette_score,
    calinski_harabasz_score,
)

# Generate half-moon dataset (500 samples, noise=0.2)
X_moons, y_moons = make_moons(n_samples=500, noise=0.2, random_state=42)

# Fit K-Means clustering
kmeans = KMeans(n_clusters=2, n_init=10, random_state=42)
y_pred = kmeans.fit_predict(X_moons)

# Evaluate cluster quality on four metrics
ari = adjusted_rand_score(y_moons, y_pred)
v_meas = v_measure_score(y_moons, y_pred)
sil = silhouette_score(X_moons, y_pred)
ch_idx = calinski_harabasz_score(X_moons, y_pred)

print("K-Means Clustering Metrics (Half-Moon Dataset)")
print(f"(i)   Adjusted Rand Index (ARI) : {ari:.4f}")
print(f"(ii)  V-measure                 : {v_meas:.4f}")
print(f"(iii) Silhouette Score          : {sil:.4f}")
print(f"(iv)  Calinski-Harabasz Index   : {ch_idx:.4f}")

# Plot true labels vs. K-Means clusters
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].scatter(X_moons[:, 0], X_moons[:, 1], c=y_moons, cmap="coolwarm", edgecolors="k", s=25)
axes[0].set_title("True Half-Moon Labels")
axes[0].grid(True, linestyle="--", alpha=0.5)

axes[1].scatter(X_moons[:, 0], X_moons[:, 1], c=y_pred, cmap="coolwarm", edgecolors="k", s=25)
axes[1].scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    c="black",
    marker="X",
    s=180,
    label="Centroids",
)
axes[1].set_title("K-Means Clustering (K=2)")
axes[1].legend()
axes[1].grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
