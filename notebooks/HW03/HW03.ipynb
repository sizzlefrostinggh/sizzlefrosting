# 3 Constrained Optimization using Lagrange Multiplier

Using method of Lagrange mulitplier, minimize $f(x, y, z) = x^2 + y^2 + z^2$ subject to $x + 2y + z = 6$.

## 1. Set up the Lagrangian.

$\mathcal{L}(x, y, z, \lambda) = x^2 + y^2 + z^2 - \lambda(x + 2y + z - 6)$

## 2. Derive the system of equations.

$\frac{\partial \mathcal{L}}{\partial x} = 2x - \lambda = 0$

$\frac{\partial \mathcal{L}}{\partial y} = 2y - 2\lambda = 0$

$\frac{\partial \mathcal{L}}{\partial z} = 2z - \lambda = 0$

$\frac{\partial \mathcal{L}}{\partial \lambda} = -(x + 2y + z - 6) = 0$

## 3. Solve for critical points.

$2x = \lambda$

$x = \frac{\lambda}{2}$

$2y = 2\lambda$

$y = \lambda$

$2z = \lambda$

$z = \frac{\lambda}{2}$

$\left(\frac{\lambda}{2}\right) + 2(\lambda) + \left(\frac{\lambda}{2}\right) = 6$

$\frac{\lambda}{2} + \frac{4\lambda}{2} + \frac{\lambda}{2} = 6$

$3\lambda = 6$

$\lambda = 2$

$x = \frac{2}{2} = 1$

$y = 2$

$z = \frac{2}{2} = 1$

The critical point is $(1, 2, 1)$.

## 4. Classify extrema.

$f(1, 2, 1) = 1^2 + 2^2 + 1^2$

$f(1, 2, 1) = 1 + 4 + 1$

$f(1, 2, 1) = 6$

We know this is a minimum because the original equation ($f(x, y, z) = x^2 + y^2 + z^2$) is an infinite paraboloid extending upwards.

## 5. Explain the role of the constraint.

The role of the constraint is that without it, the minimum would go to the bottom which would be (0, 0, 0).  With this constraint we ignore the bottom of the "bowl" of the 3D space.

# 5 Linear Regression and Regularization

$J(w) = (y - Xw)^\top (y - Xw) + \lambda w^\top w$

$J(w) = (y^\top - w^\top X\top)(y - Xw) + \lambda w^\top w$

$J(w) = y^\top y - y^\top Xw - w^\top X^\top y + w^\top X^\top Xw + \lambda w^\top w$

$J(w) = y^\top y - 2w^\top X^\top y + w^\top X^\top Xw + \lambda w^\top w$

$\nabla_{w} J(w) = 0 - 2X^\top y + 2X^\top Xw + 2\lambda w$

$- 2X^\top y + 2X^\top Xw + 2\lambda w = 0$

$X^\top Xw + \lambda w = X^\top y$

$(X^\top X + \lambda I)w = X^\top y$

$\hat{w} = (X^\top X + \lambda I)^{-1} X^\top y$

The original solution doesn't have the $\lambda I$ added to the $X^\top X$ which is the penalty.  The impact of adding this penalty will help smooth out the line.  It mathematically punishes the algorithm for assigning extreme weights to any single feature.

# 6 Assumptions in Linear Regression

The assumptions in linear regression is that we assume the relationship is a straight line.  We assume the sum of the mean of error is 0.  We assume that each data point is independent from the other and doesn't influence other data points.  Finally we assume that the data points are spread out evenly along the line.

# 7 Computing Derivative using PyTorch Autograd

$\tau  = 2 r \cdot \sqrt{\cfrac{r}{r_s}} \cdot \sqrt{1 - \cfrac{r_s}{r}}$

Compute the derivative $\cfrac{d\tau}{dr}$ considering the value of $r = 30000.1$  for $r_s = 0.00887$

```python
import torch

# Define Schwarzschild radius
r_s = torch.tensor(0.00887)
# Define r
r = torch.tensor(30000.1, requires_grad = True)
# Define tau
tau = 2 * r * torch.sqrt(r / r_s) * torch.sqrt(1 - (r_s / r))
# Backward pass to compute gradient
tau.backward()
# Derivitive is stored in grad
derivative = r.grad.item()
# Print the derivative
print(f"The derivative is: {derivative}")
```

# 8 Linear Regression in One Variable with Scikit-Learn

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Read in csv as dataframe
url = "https://raw.githubusercontent.com/rahulbhadani/Datasets/master/RAV4/2020-04-15-05-12-00_2T3Y1RFV8KC014025_CAN_Messages_5_Hz.csv"
df = pd.read_csv(url)

# Set columns to steer_angle and yaw_rate
steer_col = "steer_angle"
yaw_col = "yaw_rate"

# Drop any missing rows in steer_col and yaw_col
df = df.dropna(subset = [steer_col, yaw_col])

# Set steer_col to independent (X)
X = df[[steer_col]]
# Set yaw_col to response (y)
y = df[yaw_col]

# Set test size to 25%
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 42)

# Initialize liner regression model and fit model to data
model = LinearRegression()
model.fit(X_train, y_train)

# Evaluate the R-squared score
y_hat = model.predict(X_test)
r_squared = r2_score(y_test, y_hat)

print(f"Estimated coefficient: {model.coef_[0]}")
print(f"Estimated intercept:   {model.intercept_}")
print(f"R-squared:             {r_squared}")
```
